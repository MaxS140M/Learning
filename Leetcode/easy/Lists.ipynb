Example 1:
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].


Example 2:
Input: nums = [3,2,4], target = 6
Output: [1,2]

Example 3:
Input: nums = [3,3], target = 6
Output: [0,1]

```python
from typing import List


class Solution:
    def twoSum(self, nums: List[int], target: int) -> List[int]:
        seen = {} # Dictionary to store seen values and their indices
        for i, value in enumerate(nums): # Loop through the list with index and value
            remaining = target - nums[i] # Calculate the remaining value needed to reach the target
            
            if remaining in seen: # Check if the remaining value is already in the seen dictionary
                return [i, seen[remaining]]  # Return the current index and the index of the remaining value
            else:
                seen[value] = i  # Add the current value and its index to the seen dictionary

twosum = Solution()
result = twosum.twoSum([2,7,11,15], 9)
print(result)

result2 = twosum.twoSum([3,2,4], 6)
print(result2)

result3 = twosum.twoSum([3,3], 6)
print(result3)

result4 = twosum.twoSum([3,4], 6)
print(result4)
```
